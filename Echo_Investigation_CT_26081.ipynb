# ***Module 05 — The Damaged Message***

```python
stream = "100011001001001011100110001011000000001100011011000111000000110000001000000111001010101011111111"

def recover_message(bit_stream):
    recovered = ""
    for i in range(0, len(bit_stream), 3):
        triple = bit_stream[i:i+3]
        # Majority vote rule to recover original bit
        if triple.count('1') > triple.count('0'):
            recovered += "1"
        else:
            recovered += "0"
    return recovered

print("Recovered Message Text:", recover_message(stream))
# Note: This scheme fails if heavy noise flips 2 out of 3 bits in a triple
recovered_bin = recover_message(stream)
print("Recovered Binary Payload:", recovered_bin)

# Simple loop to convert 8-bit chunks into ASCII characters
ascii_message = ""
for i in range(0, len(recovered_bin), 8):
    byte = recovered_bin[i:i+8]       # Get 8 bits
    decimal_value = int(byte, 2)     # Convert binary to decimal number
    character = chr(decimal_value)   # Convert decimal to ASCII character
    ascii_message += character       # Add to our final message

print("Decoded Plaintext Message:", ascii_message)
```

* **Answer & Proof:** Majority voting decodes binary payload to ASCII **"ECHO"**; proves error correction and recovery of damaged transmission streams.

### **Module 05 — The Damaged Message (Mandatory)**
* **Bronze Depth:** Processed 96-bit corrupted stream across 32 triples using a 3-bit repetition majority vote algorithm. Extracted binary payload `01000101 01000011 01001000 01001111` and decoded the ASCII plaintext word: **`ECHO`**.
* **Silver Depth:** Identified failure threshold: majority vote error correction breaks when bit flip noise corrupts $\ge 2$ out of 3 bits in a triple.
